Task- 1:
Create a Python program that converts between different units of measurement.
• The program should:
1. Prompt the user to choose the type of conversion (e.g., length, weight, volume).
2. Ask the user to input the value to be converted.
3. Perform the conversion and display the result.
4. Handle potential errors, such as invalid input or unsupported conversion types.
• Requirements:
1. Functions: Define at least one function to perform the conversion.
2. Error Handling: Use try-except blocks to handle invalid input (e.g., non-numeric values).
3. User Input: Prompt the user to select the conversion type and input the value.
4. Docstrings: Include a docstring in your function to describe its purpose, parameters, and
return value.
• Conversion Options:
1. Length:– Convert meters (m) to feet (ft).– Convert feet (ft) to meters (m).
2. Weight:– Convert kilograms (kg) to pounds (lbs).– Convert pounds (lbs) to kilograms (kg).
3. Volume:– Convert liters (L) to gallons (gal).– Convert gallons (gal) to liters (L)

```python
def convert_units(conversion_type, value):
    """
    Converts units based on the selected conversion type.

    Parameters:
    conversion_type (str): The type of conversion chosen by the user (1–6).
    value (float): The numeric value to be converted.

    Returns:
    float: The converted value if valid conversion type is provided.
    None: If the conversion type is invalid.
    """

    if conversion_type == "1":  # meters to feet
        return value * 3.28084
    elif conversion_type == "2":  # feet to meters
        return value / 3.28084
    elif conversion_type == "3":  # kilograms to pounds
        return value * 2.20462
    elif conversion_type == "4":  # pounds to kilograms
        return value / 2.20462
    elif conversion_type == "5":  # liters to gallons
        return value * 0.264172
    elif conversion_type == "6":  # gallons to liters
        return value / 0.264172
    else:
        return None


def main():
    print("====== Unit Conversion Program ======")
    print("Length Conversions:")
    print("1. Meters to Feet")
    print("2. Feet to Meters")
    print("\nWeight Conversions:")
    print("3. Kilograms to Pounds")
    print("4. Pounds to Kilograms")
    print("\nVolume Conversions:")
    print("5. Liters to Gallons")
    print("6. Gallons to Liters")

    choice = input("\nChoose conversion type (1-6): ")

    try:
        value = float(input("Enter value to convert: "))

        result = convert_units(choice, value)

        if result is not None:
            print(f"\nConverted value: {result:.4f}")
        else:
            print("\nError: Invalid conversion type selected.")

    except ValueError:
        print("\nError: Please enter a valid numeric value.")


if __name__ == "__main__":
    main()
```

Create a Python program that performs various mathematical operations on a list of numbers.
• The Program should:
1. Prompt the user to choose an operation (e.g., find the sum, average, maximum, or minimum

```python
def calculate_sum(numbers):
    """
    Calculates the sum of a list of numbers.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The total sum of the numbers.
    """
    return sum(numbers)


def calculate_average(numbers):
    """
    Calculates the average of a list of numbers.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The average of the numbers.
    """
    return sum(numbers) / len(numbers)


def find_maximum(numbers):
    """
    Finds the largest number in a list.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The maximum number.
    """
    return max(numbers)


def find_minimum(numbers):
    """
    Finds the smallest number in a list.

    Parameters:
    numbers (list): A list of numeric values.

    Returns:
    float: The minimum number.
    """
    return min(numbers)


def main():
    print("Mathematical Operations Program")
    print("1. Sum")
    print("2. Average")
    print("3. Maximum")
    print("4. Minimum")

    choice = input("Choose operation (1-4): ")

    try:
        user_input = input("Enter numbers separated by spaces: ")
        numbers = user_input.split()

        # Convert input to float
        numbers = [float(num) for num in numbers]

        if len(numbers) == 0:
            print("Error: List cannot be empty.")
            return

        if choice == "1":
            result = calculate_sum(numbers)
            print("Sum:", result)

        elif choice == "2":
            result = calculate_average(numbers)
            print("Average:", result)

        elif choice == "3":
            result = find_maximum(numbers)
            print("Maximum:", result)

        elif choice == "4":
            result = find_minimum(numbers)
            print("Minimum:", result)

        else:
            print("Invalid choice.")

    except ValueError:
        print("Error: Please enter only numeric values.")


if __name__ == "__main__":
    main()
```

Write a Python function that takes a nested list and flattens it into a single list, where all the elements

• Requirements:– Define a function flatten(lst) that takes a nested list lst and returns a flattened version
of the list.– Example: For the input [[1, 2], [3, 4], [5]], the output should be [1, 2, 3, 4, 5].

```python
def flatten(lst):
    flat_list = []
    for sublist in lst:
        for item in sublist:
            flat_list.append(item)
    return flat_list
```

```python
def sum_nested(lst):
    total = 0
    for item in lst:
        if isinstance(item, list):
            total += sum_nested(item)  # recursive call for nested list
        else:
            total += item
    return total
```

4. Remove Specific Element from a Nested List:
Write a Python function that removes all occurrences of a specific element from a nested list.
• Requirements:– Define a function remove

modified list.– Example: Fortheinput lst = [[1, 2], [3, 2], [4, 5]]andelem = 2,theoutputshould

```python
def remove_element(lst, elem):
    """
    Removes all occurrences of elem from a nested list.

    Parameters:
    lst (list): A nested list (list of lists).
    elem: The element to remove.

    Returns:
    list: The modified nested list with elem removed.
    """
    new_list = []

    for sublist in lst:
        filtered = []
        for item in sublist:
            if item != elem:
                filtered.append(item)
        new_list.append(filtered)

    return new_list


# Example usage
nested = [[1, 2], [3, 2], [4, 5]]
elem = 2

result = remove_element(nested, elem)

print("Original nested list:", nested)
print("Nested list after removing", elem, ":", result)
```

5. Find the Maximum Element in a Nested List:
Write a Python function that finds the maximum element in a nested list (regardless of depth).
• Requirements:– Define a function find
element.
max(lst) that takes a nested list lst and returns the maximum– Example: For the input [[1, 2], [3, [4, 5]], 6], the output should be 6.

```python
def find_element_max(lst):
    """
    Finds the maximum element in a nested list,
    regardless of how deeply nested it is.

    Parameters:
    lst (list): A nested list containing numbers.

    Returns:
    int/float: The maximum element in the nested list.
    """

    max_val = float('-inf')

    for item in lst:
        if isinstance(item, list):
            nested_max = find_element_max(item)
            if nested_max > max_val:
                max_val = nested_max
        else:
            if item > max_val:
                max_val = item

    return max_val


# Example usage
nested = [[1, 2], [3, [4, 5]], 6]

result = find_element_max(nested)

print("Nested list:", nested)
print("Maximum element:", result)
```

6. Count Occurrences of an Element in a Nested List:
Write a Python function that counts how many times a specific element appears in a nested list.

occurrences(lst, elem) that counts the occurrences of elem in
the nested list lst.– Example: Fortheinput lst = [[1, 2], [2, 3], [2, 4]]andelem = 2,theoutputshould
be 3

```python
def count_occurrences(lst, elem):
    count = 0
    for item in lst:
        if isinstance(item, list):
            # Recursively count in nested list
            count += count_occurrences(item, elem)
        else:
            if item == elem:
                count += 1
    return count
```

7. Flatten a List of Lists of Lists:
Write a Python function that flattens a list of lists of lists into a single list, regardless of the depth.
• Requirements:– Define a function deep
flatten(lst) that takes a deeply nested list lst and returns a single
f
lattened list.– Example: For the input [[[1, 2], [3, 4]], [[5, 6], [7, 8]]], the output should be
[1, 2, 3, 4, 5, 6, 7, 8].

```python
def deep_flatten(lst):
    """
    Flattens a deeply nested list into a single list,
    regardless of how many levels it contains.

    Parameters:
    lst (list): A deeply nested list.

    Returns:
    list: A single flattened list containing all elements.
    """
    flat_list = []

    for item in lst:
        if isinstance(item, list):
            flat_list.extend(deep_flatten(item))
        else:
            flat_list.append(item)

    return flat_list


# Example usage
nested = [[[1, 2], [3, 4]], [[5, 6], [7, 8]]]

result = deep_flatten(nested)

print("Original deeply nested list:", nested)
print("Flattened list:", result)
```

8. Nested List Average:
Write a Python function that calculates the average of all elements in a nested list.
• Requirements:– Define a function average
of all the elements.
nested(lst) that takes a nested list lst and returns the average– Example: For the input [[1, 2], [3, 4], [5, 6]], the output should be 3.5

```python
def average_nested(lst):
    def helper(sublist):
        total = 0
        count = 0
        for item in sublist:
            if isinstance(item, list):
                t, c = helper(item)  # recursively get sum and count
                total += t
                count += c
            else:
                total += item
                count += 1
        return total, count

    total_sum, total_count = helper(lst)
    if total_count == 0:
        return 0  # Avoid division by zero
    return total_sum / total_count
```

```python
nested = [[1, 2], [3, 4], [5, 6]]

result = average_nested(nested)

print("Nested list:", nested)
print("Average of all elements:", result)
```

```python
#Element-wise addition using Python lists
start_time = time.time()

# Using list comprehension for element-wise addition
result_list = [list1[i] + list2[i] for i in range(N)]

end_time = time.time()
print("Time taken using Python lists: {:.6f} seconds".format(end_time - start_time))
```

```python
#Element-wise addition using NumPy arrays
start_time = time.time()

# Using NumPy vectorized addition
result_array = array1 + array2

end_time = time.time()
print("Time taken using NumPy arrays: {:.6f} seconds".format(end_time - start_time))
```

NumPy is significantly faster than regular Python lists when performing large element-wise operations.

This is because NumPy relies on optimized, vectorized implementations written in C, whereas Python lists handle operations one element at a time using Python code, which makes them slower.

```python
#Create lists and arrays
N = 1_000_000

# Python lists
list1 = list(range(N))
list2 = list(range(N))

# NumPy arrays
array1 = np.arange(N)
array2 = np.arange(N)
```

```python
# Element-wise multiplication using Python lists
start_time = time.time()

# List comprehension for element-wise multiplication
result_list = [list1[i] * list2[i] for i in range(N)]

end_time = time.time()
print("Time taken using Python lists (multiplication): {:.6f} seconds".format(end_time - start_time))
```

3. Dot Product
• Using Python Lists, compute the dot product of two lists of size 1,000,000. Measure and

Print the time taken for this operation.
• Using Numpy Arrays, Repeat the calculation and measure and print the time taken for
this operation.

```python
# Create lists and arrays
N = 1_000_000

# Python lists
list1 = list(range(N))
list2 = list(range(N))

# NumPy arrays
array1 = np.arange(N)
array2 = np.arange(N)
```

```python
# Dot product using Python lists
start_time = time.time()

# Using sum of element-wise multiplication
dot_list = sum(list1[i] * list2[i] for i in range(N))

end_time = time.time()
print("Time taken using Python lists (dot product): {:.6f} seconds".format(end_time - start_time))
```

```python
# Dot product using NumPy arrays
start_time = time.time()

# NumPy dot product
dot_array = np.dot(array1, array2)

end_time = time.time()
print("Time taken using NumPy arrays (dot product): {:.6f} seconds".format(end_time - start_time))
```

4. Matrix Multiplication
• Using Python lists, perform matrix multiplication of two matrices of size 1000x1000. Mea
sure and print the time taken for this operation.

• Using NumPy arrays, perform matrix multiplication of two matrices of size 1000x1000.
Measure and print the time taken for this operation.

```python
# Create matrices
N = 1000

# Python lists (nested lists)
list1 = [[i + j for j in range(N)] for i in range(N)]
list2 = [[i - j for j in range(N)] for i in range(N)]

# NumPy arrays
array1 = np.array(list1)
array2 = np.array(list2)
```

```python
# Matrix multiplication using Python lists
start_time = time.time()

# Initialize result matrix
result_list = [[0 for _ in range(N)] for _ in range(N)]

# Naive nested loop multiplication
for i in range(N):
    for j in range(N):
        for k in range(N):
            result_list[i][j] += list1[i][k] * list2[k][j]

end_time = time.time()
print("Time taken using Python lists (matrix multiplication): {:.6f} seconds".format(end_time - start_time))
```

```python
#Matrix multiplication using NumPy arrays
start_time = time.time()

# NumPy matrix multiplication (vectorized)
result_array = np.dot(array1, array2)

end_time = time.time()
print("Time taken using NumPy arrays (matrix multiplication): {:.6f} seconds".format(end_time - start_time))
```
